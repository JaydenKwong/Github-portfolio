# 4.3 Feature Engineering

4.3.1 Theme 1 – Provider Operational Behaviour
Feature 1 – Total Inpatient Claims

```python
# Engineer Feature 1: Calculate the total number of inpatient claims submitted by each healthcare provider.

inpatient[['Provider', 'ClaimID']].head(10)
```

```python
# Calculate the total number of inpatient claims submitted by each healthcare provider.

total_inpatient_claims = (
    inpatient
    .groupby('Provider')['ClaimID']
    .count()
    .reset_index(name='Total_Inpatient_Claims')
)
```

```python
# Engineer Feature 2: Calculate the total number of outpatient claims submitted by each healthcare provider.

total_outpatient_claims = (
    outpatient
    .groupby('Provider')['ClaimID']
    .count()
    .reset_index(name='Total_Outpatient_Claims')
)
```

# ============================================================
# Feature 3 – Total Claims
# Theme 1 – Provider Claim Behaviour
# ============================================================

```python
# Display the first few records of the inpatient and outpatient claim features before calculating total claims.

total_inpatient_claims.head()
```

```python
total_outpatient_claims.head()
```

```python
# Merge the inpatient and outpatient claim counts temporarily to calculate total claims.

total_claims_temp = pd.merge(
    total_inpatient_claims,
    total_outpatient_claims,
    on='Provider',
    how='outer'
)
```

```python
# Replace missing claim counts with zero for providers without inpatient or outpatient claims.

total_claims_temp[['Total_Inpatient_Claims', 'Total_Outpatient_Claims']] = (
    total_claims_temp[['Total_Inpatient_Claims', 'Total_Outpatient_Claims']]
    .fillna(0)
)
```

```python
# Engineer Feature 3: Calculate the total number of claims submitted by each healthcare provider.

total_claims_temp['Total_Claims'] = (
    total_claims_temp['Total_Inpatient_Claims'] +
    total_claims_temp['Total_Outpatient_Claims']
)
```

```python
# Keep only the Provider and engineered Total_Claims feature for later merging.

total_claims = total_claims_temp[['Provider', 'Total_Claims']]
```

```python
# Preview the engineered Total_Claims feature.

total_claims.head()
```

```python
# Verify the number of providers after calculating total claims.

print(total_claims.shape)
```

# ============================================================
# Feature 4 – Number of Unique Beneficiaries
# Theme 1 – Provider Claim Behaviour
# ============================================================

```python
# Display the Provider and BeneID columns from inpatient and outpatient claims used to identify unique beneficiaries.

inpatient[['Provider', 'BeneID']].head(10)
```

```python
outpatient[['Provider', 'BeneID']].head(10)
```

```python
# Combine inpatient and outpatient Provider-BeneID records before counting unique beneficiaries.

beneficiary_claims_temp = pd.concat(
    [
        inpatient[['Provider', 'BeneID']],
        outpatient[['Provider', 'BeneID']]
    ],
    ignore_index=True
)
```

```python
# Engineer Feature 4: Calculate the number of unique beneficiaries treated by each healthcare provider.

unique_beneficiaries = (
    beneficiary_claims_temp
    .groupby('Provider')['BeneID']
    .nunique()
    .reset_index(name='Unique_Beneficiaries')
)
```

```python
# Preview the engineered Unique_Beneficiaries feature.

unique_beneficiaries.head()
```

```python
# Verify the number of providers after calculating the number of unique beneficiaries.

print(unique_beneficiaries.shape)
```

Merge First 4 features into Theme 1

```python
# Merge the first four Theme 1 claim activity features using Provider as the common key.

theme1_features = (
    total_inpatient_claims
    .merge(total_outpatient_claims, on='Provider', how='outer')
    .merge(total_claims, on='Provider', how='outer')
    .merge(unique_beneficiaries, on='Provider', how='outer')
)
```

```python
# Replace missing claim activity values with zero for providers without a specific claim type.

theme1_features = theme1_features.fillna(0)
```

```python
# Preview the Theme 1 feature dataset after combining the first four engineered features.

theme1_features.head()
```

```python
# Verify the dimensions of the Theme 1 feature dataset.

print(theme1_features.shape)
```

```python
# Convert beneficiary date of birth into datetime format

beneficiary['DOB'] = pd.to_datetime(beneficiary['DOB'])
```
