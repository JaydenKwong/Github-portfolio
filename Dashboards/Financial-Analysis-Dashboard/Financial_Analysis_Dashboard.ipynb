```python
# Define the core financial columns used for the dashboard
core_columns = [
    'grossProfit',
    'operatingIncome',
    'operatingExpenses',
    'netIncome'
]

# Display rows containing missing values in the core financial columns
df[df[core_columns].isnull().any(axis=1)][
    ['symbol', 'fiscalDateEnding'] + core_columns
]
```

```python
# Remove rows with missing values in the core financial columns
df = df.dropna(subset=core_columns)

# Reset the row index after removing the incomplete records
df = df.reset_index(drop=True)

# Check the updated dataset shape
print("Updated shape:", df.shape)
```

```python
# Select the variables needed for the financial analysis dashboard
financial_df = df[[
    'symbol',
    'fiscalDateEnding',
    'reportedCurrency',
    'totalRevenue',
    'costOfRevenue',
    'grossProfit',
    'operatingExpenses',
    'operatingIncome',
    'researchAndDevelopment',
    'sellingGeneralAndAdministrative',
    'netIncome'
]].copy()

# Preview the selected financial data
financial_df.head()
```

```python
# Calculate gross profit margin as a percentage of total revenue
financial_df['grossMarginPct'] = (
    financial_df['grossProfit'] / financial_df['totalRevenue']
) * 100

# Preview the new gross margin variable
financial_df[
    ['symbol', 'fiscalDateEnding', 'totalRevenue',
     'grossProfit', 'grossMarginPct']
].head()
```

```python
# Sort each company's financial records from oldest to newest
financial_df = financial_df.sort_values(
    ['symbol', 'fiscalDateEnding']
).reset_index(drop=True)

# Preview NVIDIA dates to verify chronological order
financial_df[
    financial_df['symbol'] == 'NVDA'
][['symbol', 'fiscalDateEnding', 'totalRevenue']].head(10)
```

```python
# Calculate gross margin percentage
financial_df['grossMarginPct'] = (
    financial_df['grossProfit'] / financial_df['totalRevenue']
) * 100

# Calculate operating margin percentage
financial_df['operatingMarginPct'] = (
    financial_df['operatingIncome'] / financial_df['totalRevenue']
) * 100

# Calculate net profit margin percentage
financial_df['netProfitMarginPct'] = (
    financial_df['netIncome'] / financial_df['totalRevenue']
) * 100

# Calculate operating expense ratio as a percentage of revenue
financial_df['operatingExpenseRatioPct'] = (
    financial_df['operatingExpenses'] / financial_df['totalRevenue']
) * 100
```

```python
# Calculate quarter-over-quarter revenue growth percentage
financial_df['revenueGrowthQoQPct'] = (
    financial_df.groupby('symbol')['totalRevenue']
    .pct_change() * 100
)

# Calculate year-over-year revenue growth percentage
financial_df['revenueGrowthYoYPct'] = (
    financial_df.groupby('symbol')['totalRevenue']
    .pct_change(periods=4) * 100
)

# Calculate quarter-over-quarter operating expense growth percentage
financial_df['operatingExpenseGrowthQoQPct'] = (
    financial_df.groupby('symbol')['operatingExpenses']
    .pct_change() * 100
)

# Calculate quarter-over-quarter net income growth percentage
financial_df['netIncomeGrowthQoQPct'] = (
    financial_df.groupby('symbol')['netIncome']
    .pct_change() * 100
)
```
