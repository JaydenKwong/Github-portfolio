```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/tech_advertising_campaigns_dataset.csv')
```

```python
# Find rows containing at least one empty/blank cell
df[df.astype(str).apply(lambda x: x.str.strip() == '').any(axis=1)]
```

```python
df.duplicated().sum()
```

```python
# Convert start_date from text to datetime
df['start_date'] = pd.to_datetime(df['start_date'])
```

```python
# Verify whether the provided CTR matches our own calculation
df['CTR_Check'] = (df['clicks'] / df['impressions']) * 100

# Compare the provided CTR with our calculated CTR
(df[['CTR', 'CTR_Check']].head(10)).round(3)
```

```python
# Calculate profit margin only when revenue is greater than zero
df['profit_margin'] = np.where(
    df['revenue'] > 0,
    (df['profit'] / df['revenue']) * 100,
    np.nan
)

# Check the profit margin results
df[['revenue', 'profit', 'profit_margin']].head()
```

```python
# Check how many campaigns have zero revenue
(df['revenue'] == 0).sum()
```

```python
# Remove the temporary CTR verification variable
df.drop(columns=['CTR_Check'], inplace=True)

# Check the final dataset
df.info()
```

```python
# Export the cleaned dataset for Tableau
df.to_csv('cleaned_digital_advertising_campaigns.csv', index=False)

# Download the cleaned dataset
from google.colab import files
files.download('cleaned_digital_advertising_campaigns.csv')
```
