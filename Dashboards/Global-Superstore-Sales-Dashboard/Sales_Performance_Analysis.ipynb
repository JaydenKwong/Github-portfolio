```python
# Find sales values containing characters other than numbers or decimal points

df[df['sales'].str.contains(r'[^0-9.]', regex=True)]['sales'].head(20)
```

```python
# Remove commas and convert sales to numeric

df['sales'] = df['sales'].str.replace(',', '', regex=False)
df['sales'] = pd.to_numeric(df['sales'])
```

```python
# Convert order and shipping dates from DD/MM/YYYY to datetime

df['order_date'] = pd.to_datetime(df['order_date'], format='%d/%m/%Y')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='%d/%m/%Y')
```

```python
# Verify the converted date columns

df[['order_date', 'ship_date']].head()
```

```python
# Verify updated data types

df[['order_date', 'ship_date', 'sales']].dtypes
```

```python
# Review unique values in important categorical columns

for column in ['ship_mode', 'segment', 'market', 'region',
               'category', 'sub_category', 'order_priority']:
    print(f'\n{column}:')
    print(df[column].unique())
```

```python
# Create time-based fields for analysis

df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_month_name'] = df['order_date'].dt.month_name()
df['year_month'] = df['order_date'].dt.to_period('M')
```

```python
# Check whether the original year matches the order date

(df['year'] == df['order_year']).value_counts()
```

```python
# Calculate overall sales performance KPIs

total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
total_orders = df['order_id'].nunique()
total_customers = df['customer_name'].nunique()
profit_margin = (total_profit / total_sales) * 100

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
```

```python
# Remove duplicate year field after validation

df.drop(columns=['order_year'], inplace=True)
```

```python
# Save the cleaned dataset

df.to_csv('/content/superstore_sales_cleaned.csv', index=False)
```
